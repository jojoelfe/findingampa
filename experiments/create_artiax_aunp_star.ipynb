```python
import os 

def create_artiax_aunp_starfile():

    import starfile

    data = starfile.read("aunps/aunp_tm_BP.star")
    print(data.columns)
    data['rlnCoordinateX'] = data['faTM_PeakX'] + data['faTomo_gaussianFitX']
    data['rlnCoordinateY'] = data['faTM_PeakY'] + data['faTomo_gaussianFitY']
    data['rlnCoordinateZ'] = data['faTM_PeakZ'] + data['faTomo_gaussianFitZ']

    starfile.write( data, "aunps/aunp_artiax.star", overwrite=True)
```

```python
with open("/goliath/processing/Gouaux/CJS/BestTomo/ProcessingCJS/tomograms/15F1_tomograms/TOP_TOMOS/tomo_for_sta.txt","r") as f:
    tomos_for_sta = f.readlines()

tomos_for_sta = ["20240209_waffleHip_165"]
for tomo in tomos_for_sta:
    tomo = tomo.strip()
    print(tomo)
    os.chdir(f'/goliath/processing/Gouaux/CJS/BestTomo/ProcessingCJS/tomograms/15F1_tomograms/TOP_TOMOS/{tomo}/best_alignment/')
    create_artiax_aunp_starfile()
```
